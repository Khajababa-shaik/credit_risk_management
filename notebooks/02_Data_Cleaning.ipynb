```python
# ==========================================================
# Import Required Libraries
# ==========================================================

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
import missingno as msno

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)

print("=" * 60)
print("Libraries Imported Successfully")
print("=" * 60)
```

```python
# ==========================================================
# Load Dataset
# ==========================================================

DATA_PATH = "../data/raw/german_credit_data.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset Loaded Successfully")
```

```python
# ==========================================================
# Create Backup Copy
# ==========================================================

clean_df = df.copy()

print("Backup Copy Created Successfully")
```

```python
# ==========================================================
# Dataset Shape
# ==========================================================

print(f"Rows    : {clean_df.shape[0]}")
print(f"Columns : {clean_df.shape[1]}")
```

```python
# ==========================================================
# Missing Value Analysis
# ==========================================================

missing = pd.DataFrame({

    "Missing Values": clean_df.isnull().sum(),

    "Percentage": round(
        clean_df.isnull().mean() * 100,
        2
    )

})

missing = missing.sort_values(

    by="Missing Values",

    ascending=False

)

missing
```

```python
# ==========================================================
# Missing Value Analysis
# ==========================================================

missing = pd.DataFrame({
    "Missing Values": clean_df.isnull().sum(),
    "Percentage (%)": round((clean_df.isnull().sum() / len(clean_df)) * 100, 2)
})

missing = missing[missing["Missing Values"] > 0]

missing.sort_values(
    by="Missing Values",
    ascending=False
)
```

```python
# ==========================================================
# Handle Missing Values
# ==========================================================

clean_df["Saving accounts"] = clean_df["Saving accounts"].fillna("No Account")

clean_df["Checking account"] = clean_df["Checking account"].fillna("No Account")

print("=" * 60)
print("Missing Values Handled Successfully")
print("=" * 60)
```

```python
clean_df.isnull().sum()
```

```python
"""Interview Question

Q: Why didn't you use the mode?

Answer:

Missing values in "Saving accounts" and "Checking account" represent customers who likely do not have those account types. Replacing them with the mode would incorrectly assign an account category. Creating a new category ("No Account") preserves useful business information and avoids misleading the model."""
```

```python
# ==========================================================
# Duplicate Record Analysis
# ==========================================================

duplicate_count = clean_df.duplicated().sum()

print("=" * 60)
print(f"Total Duplicate Records : {duplicate_count}")
print("=" * 60)
```

```python
# ==========================================================
# Remove Duplicate Records
# ==========================================================

rows_before = clean_df.shape[0]

clean_df = clean_df.drop_duplicates()

rows_after = clean_df.shape[0]

print("=" * 60)
print(f"Rows Before Cleaning : {rows_before}")
print(f"Rows After Cleaning  : {rows_after}")
print(f"Duplicates Removed   : {rows_before - rows_after}")
print("=" * 60)
```

```python
"""Interview Question

Q: Why remove duplicate rows?

Answer:

Duplicate records can cause the model to learn repeated patterns, which may bias predictions and reduce its ability to generalize to unseen data. Removing duplicates helps create a more reliable training dataset."""
```

```python
# ==========================================================
# Data Type Validation
# ==========================================================

validation = pd.DataFrame({
    "Column": clean_df.columns,
    "Data Type": clean_df.dtypes.astype(str),
    "Missing Values": clean_df.isnull().sum().values,
    "Unique Values": clean_df.nunique().values
})

validation
```

```python
# ==========================================================
# Numerical Feature Summary
# ==========================================================

numerical_columns = clean_df.select_dtypes(include=["int64", "float64"]).columns

clean_df[numerical_columns].describe().T
```

```python
# ==========================================================
# Categorical Feature Summary
# ==========================================================

categorical_columns = clean_df.select_dtypes(include=["object"]).columns

for column in categorical_columns:
    print("\n" + "=" * 60)
    print(f"{column}")
    print("=" * 60)
    print(clean_df[column].value_counts())
```

```python
# ==========================================================
# Save Cleaned Dataset
# ==========================================================

from pathlib import Path

processed_folder = Path("../data/processed")
processed_folder.mkdir(parents=True, exist_ok=True)

output_path = processed_folder / "cleaned_credit_data.csv"

clean_df.to_csv(output_path, index=False)

print("=" * 60)
print("Cleaned Dataset Saved Successfully")
print(f"Location : {output_path}")
print("=" * 60)
```
