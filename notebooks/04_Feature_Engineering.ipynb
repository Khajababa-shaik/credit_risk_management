```python
# ==========================================================
# Import Required Libraries
# ==========================================================

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

from pathlib import Path

import joblib

from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler,
    LabelEncoder
)

from sklearn.compose import ColumnTransformer

print("Libraries Imported Successfully")
```

```python
# ==========================================================
# Load Cleaned Dataset
# ==========================================================

DATA_PATH = "../data/processed/cleaned_credit_data.csv"

df = pd.read_csv(DATA_PATH)

print(df.shape)

df.head()
```

```python
# ==========================================================
# Dataset Information
# ==========================================================

print(df.info())

print()

print(df.isnull().sum())
```

```python
# ==========================================================
# Backup Copy
# ==========================================================

feature_df = df.copy()

print("Backup Created")
```

```python
# ==========================================================
# Features & Target
# ==========================================================

X = feature_df.drop("Risk", axis=1)

y = feature_df["Risk"]

print("Features Shape :", X.shape)
print("Target Shape :", y.shape)
```

```python
# ==========================================================
# Create Engineered Features
# ==========================================================

X["Credit_Per_Month"] = (
    X["Credit amount"] /
    X["Duration"]
)

X["Credit_Age_Ratio"] = (
    X["Credit amount"] /
    X["Age"]
)

X["Age_Duration_Product"] = (
    X["Age"] *
    X["Duration"]
)

X["Credit_Age_Product"] = (
    X["Credit amount"] *
    X["Age"]
)

X["Log_Credit_Amount"] = np.log1p(
    X["Credit amount"]
)

X["Log_Duration"] = np.log1p(
    X["Duration"]
)

X.head()
```

```python
# ==========================================================
# Numerical & Categorical Features
# ==========================================================

numerical_features = X.select_dtypes(
    include=["int64","float64"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include="object"
).columns.tolist()

print("Numerical Features")
print(numerical_features)

print()

print("Categorical Features")
print(categorical_features)
```

```python
# ==========================================================
# Column Transformer
# ==========================================================

preprocessor = ColumnTransformer(

    transformers=[

        (

            "num",

            StandardScaler(),

            numerical_features

        ),

        (

            "cat",

            OneHotEncoder(

                handle_unknown="ignore",

                sparse_output=False

            ),

            categorical_features

        )

    ],

    remainder="drop"

)

print("Preprocessor Created Successfully")
```

```python
"""Why this is better:

Automatically scales only numeric columns.
Automatically encodes only categorical columns.
Handles unseen categories during prediction.
Keeps preprocessing consistent."""
```

```python
# ==========================================================
# Transform Dataset
# ==========================================================

X_processed = preprocessor.fit_transform(X)

print(X_processed.shape)
```

```python
# ==========================================================
# Feature Names
# ==========================================================

feature_names = preprocessor.get_feature_names_out()

X_processed = pd.DataFrame(

    X_processed,

    columns=feature_names

)

X_processed.head()
```

```python
# ==========================================================
# Encode Target
# ==========================================================

target_encoder = LabelEncoder()

y = target_encoder.fit_transform(y)

print(target_encoder.classes_)
```

```python
# ==========================================================
# Save Preprocessor
# ==========================================================

MODELS = Path("../models")

MODELS.mkdir(exist_ok=True)

joblib.dump(

    preprocessor,

    MODELS / "preprocessor.pkl"

)

print("Preprocessor Saved")
```

```python
joblib.dump(

    target_encoder,

    MODELS / "target_encoder.pkl"

)

print("Target Encoder Saved")
```

```python
joblib.dump(

    feature_names,

    MODELS / "feature_columns.pkl"

)

print("Feature Names Saved")
```

```python
# ==========================================================
# Save Engineered Dataset
# ==========================================================

engineered = X_processed.copy()

engineered["Risk"] = y

engineered.to_csv(

    "../data/processed/engineered_credit_data.csv",

    index=False

)

print("Engineered Dataset Saved Successfully")
```
